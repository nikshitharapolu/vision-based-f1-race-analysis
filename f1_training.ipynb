```python
PROJECT = "/content/drive/Shareddrives/682_project/vision-based-f1-race-analysis"
LAST_PT = f"{PROJECT}/runs/f1_detector_v3/weights/last.pt"

from ultralytics import YOLO

model = YOLO(LAST_PT) 

model.train(
    data    = "/content/roboflow_dataset/data.yaml",
    epochs  = 100,
    batch   = 16,
    imgsz   = 640,
    device  = "cuda",
    project = f"{PROJECT}/runs",
    name    = "f1_detector_v3",
    resume  = True,
)
```

```python
import shutil

best = f"{PROJECT}/runs/f1_detector_v3/weights/best.pt"
dest = f"{PROJECT}/models/car_detector.pt"

shutil.copy2(best, dest)
print(f"Saved → {dest}")
```
